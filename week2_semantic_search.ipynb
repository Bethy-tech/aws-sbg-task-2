```python
# Check the dataset
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

# Clean the text
df["question"] = df["question"].str.strip()
df["answer"] = df["answer"].str.strip()

# Remove missing or empty questions and answers
df = df.dropna(subset=["question", "answer"])
df = df[
    (df["question"] != "") &
    (df["answer"] != "")
]

# Remove duplicate questions
df = df.drop_duplicates(subset=["question"])

# Reset row numbers
df = df.reset_index(drop=True)

print("\nAfter cleaning:")
print("Remaining FAQs:", len(df))

df.head()
```

```python
# Create text chunks from our FAQs

chunks = []

for _, row in df.iterrows():
    chunk = f"Question: {row['question']}\nAnswer: {row['answer']}"

    chunks.append({
        "text": chunk,
        "question": row["question"],
        "answer": row["answer"]
    })

print("Total chunks created:", len(chunks))

print("\nExample chunk:\n")
print(chunks[0]["text"])
```

```python
import chromadb
from sentence_transformers import SentenceTransformer

print("ChromaDB:", chromadb.__version__)
print("Both libraries imported successfully!")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
test_text = "Machine learning helps computers learn from data."

embedding = model.encode(test_text)

print("Embedding created successfully!")
print("Vector dimensions:", embedding.shape)
print("First 10 numbers:", embedding[:10])
```

```python
import chromadb

# Create an in-memory vector database
client = chromadb.Client()

# Safely get or create the collection for our AI/ML FAQs
collection = client.get_or_create_collection(
    name="ai_ml_faqs"
)

print("ChromaDB collection loaded successfully!")
```

```python
# Extract text and metadata from our chunks
documents = [chunk["text"] for chunk in chunks]

metadatas = [
    {
        "question": chunk["question"],
        "answer": chunk["answer"]
    }
    for chunk in chunks
]

# Generate embeddings for all FAQ chunks
embeddings = model.encode(documents).tolist()

# Store everything in ChromaDB
collection.add(
    ids=[f"faq_{i}" for i in range(len(chunks))],
    documents=documents,
    embeddings=embeddings,
    metadatas=metadatas
)

print("FAQs added to ChromaDB!")
print("Total stored:", collection.count())
```

```python
def semantic_search(query, top_k=3):
    # Convert the user's question into an embedding
    query_embedding = model.encode(query).tolist()

    # Find the most similar FAQ chunks
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )

    # Display the results
    for i, metadata in enumerate(results["metadatas"][0]):
        print(f"\nResult {i + 1}")
        print("Question:", metadata["question"])
        print("Answer:", metadata["answer"])

semantic_search("How can computers learn from examples?")
```

```python
def semantic_search(query, top_k=3):
    query_embedding = model.encode(query).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=min(top_k, collection.count()),
        include=["metadatas", "distances"]
    )

    for i, metadata in enumerate(results["metadatas"][0]):
        distance = results["distances"][0][i]

        print(f"\nResult {i + 1}")
        print("Question:", metadata["question"])
        print("Answer:", metadata["answer"])
        print("Distance:", round(distance, 4))
```
